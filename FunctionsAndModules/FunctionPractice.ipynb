# Exercise 1
Implement a program that receives the 3 sides of a triangle and calculates its area.

To calculate an area using the 3 side we need Heron's formula.
image.png


```python
# Ask for three inputs and use heron's formula to calculate the area
import math
sideA = float(input("What is side A?"))
sideB = float(input("What is side B?"))
sideC = float(input("What is side C?"))

s = (sideA + sideB + sideC) / 2

area = math.sqrt(s * (s-sideA) * (s-sideB) * (s - sideC))

print(area)
```

```python
# Implement a version that solicites 2 triangle info and return which is the
# largest. Say T1 if it's the first one and T2 if it's the second one.
import math
sideA = float(input("What is side A for T1?"))
sideB = float(input("What is side B for T1?"))
sideC = float(input("What is side C for T1?"))

s = (sideA + sideB + sideC) / 2

area1 = math.sqrt(s * (s-sideA) * (s-sideB) * (s - sideC))

# Triangle 2
sideA = float(input("What is side A for T2?"))
sideB = float(input("What is side B for T2?"))
sideC = float(input("What is side C for T2?"))

s = (sideA + sideB + sideC) / 2

area2 = math.sqrt(s * (s-sideA) * (s-sideB) * (s - sideC))

print(f"T1 area: {area1}\nT2 area:{area2}")
if area1 > area2:
    print("T1")
else:
    print("T2")
```

## Modulization!
What if the repeated code is called using a function instead?

```python
# Create a function that asks for the sides of a triangle and calculates the
# area if the triangles
import math

def calculate_triangle_area():
    sideA = float(input("What is side A?"))
    sideB = float(input("What is side B?"))
    sideC = float(input("What is side C?"))

    s = (sideA + sideB + sideC) / 2

    area1 = math.sqrt(s * (s-sideA) * (s-sideB) * (s - sideC))
    return area1

print("For trinagle 1")
area1 = calculate_triangle_area()

# Triangle 2
print("For triangle 2")
area2 = calculate_triangle_area()

print(f"T1 area: {area1}\nT2 area: {area2}")
if area1 > area2:
    print("T1")
else:
    print("T2")
```

Implement a function that checks whether the size of a plot of land complies with a minimum area requirement.

The function receives the following parameters:
- Length of the plot
- Width of the plot
- Minimum required area

The function should calculate the area of the plot and return:
- true if the area is greater than or equal to the minimum required area.
- false otherwise.

```python
def check_if_valid_plot(length, width, minimum_area):
    area = length * width
    return area >= minimum_area


check_if_valid_plot(10, 15, 50)
```

## Now I'm just showing off
This is how you can graph functions in Python.
To run it locally, your environment need the matplotlib and numpy module.

```python
import matplotlib.pyplot as plt
import numpy as np

# Generate x values
x = np.linspace(0.1, 10, 400) # Start from 0.1 to avoid log(0)

# Linear function: y = mx + b
y_linear = 2 * x + 1

# Quadratic function: y = ax^2 + bx + c
y_quadratic = 0.5 * x**2 - 2 * x + 3

# Logarithmic function: y = a * log(x) + b
y_logarithmic = 2 * np.log(x) + 1

# Create the plot
plt.figure(figsize=(10, 6))

plt.plot(x, y_linear, label='Linear Function (y = 2x + 1)', color='blue')
plt.plot(x, y_quadratic, label='Quadratic Function (y = 0.5x^2 - 2x + 3)', color='red')
plt.plot(x, y_logarithmic, label='Logarithmic Function (y = 2 log(x) + 1)', color='green')

plt.title('Comparison of Linear, Quadratic, and Logarithmic Functions')
plt.xlabel('x')
plt.ylabel('y')
plt.legend()
plt.grid(True)
plt.axhline(0, color='black', linewidth=0.5)
plt.axvline(0, color='black', linewidth=0.5)
plt.show()
```
